Implement the partial Fourier sine series for a given terms N and plot the approximation

```python
import numpy as np
import matplotlib.pyplot as plt

def func(N, l, x):
    n = np.arange(1, N + 1).reshape(-1, 1)     
    x = np.reshape(x, (1, -1))                  
    terms = np.sin(n * np.pi * x / l) / n     
    a = np.sum(terms, axis=0)                   
    return (4 / np.pi) * a

L = 50
theta = np.linspace(0, 2 * np.pi, 100)
z = func(10, L, theta)
print(z)
plt.plot(theta, z)
plt.xlabel('theta')
plt.ylabel('func value')
plt.title('Plot of func vs theta')
plt.show()
```

```python
import random 
from numpy import random

def dist(x,y,x1,y1):
    d = np.sqrt((x - x1)**2 + 2**(y - y1))
    return d
        
def nearest_neighbor(a, c):
        #print(a,c)
        f = np.min(dist(a,c,0.6962,0.5678))
        i = np.argmin(dist(a,c,0.6962,0.5678))
        return f,i

a = random.rand(100)
c = random.rand(100)
d, i = nearest_neighbor(a, c) 
print(d, i)
print(a[i], c[i])


        
        
        
```
